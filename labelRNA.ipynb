This will be the notebook for looping through the folder of 8-bit RNA mask files generated by Ryan, Liwei
and turning them 32-bit label files. So far have had good results using Voroni_Otsu labeling with spot sigma=1

```python

from tkinter import Tk
from tkinter import filedialog
from skimage.io import imread, imsave, imshow
import matplotlib
import numpy as np
import os

import pyclesperanto_prototype as cle
cle.select_device("Apple M1 Max")
```

```python
root = Tk()
root.withdraw()
#strtPath = "/Users/nelsonlab/Desktop/Test Folder for Segmentation/cell"
strtPath = "/Users/nelsonlab/"
dirName = filedialog.askdirectory(initialdir=strtPath)
rnadir = dirName + "/rna"
labPath = dirName + "/rna/rLabs"
if not os.path.exists(labPath):  
   os.mkdir(labPath)  
files = os.listdir(rnadir)
```

something is crashing and not sure what. It may be a memory problem, given the large size of the stacks (135M pixels). The voronoi-otsu labeling algorithm is supposed to require isometric voxels, but resampling doesn't seem to fix the problem and earlier it seemed to work well without resampling. 

I think I will try calculting the label image stacks in tiles and see if this fixes things. If that works, I can try to figure out how to correct any objects that get split by tile borders. 

```python
for filename in files:
   if filename.endswith(".tif"):
      print (filename)
      rPath = os.path.join(rnadir, filename)
      rnaIn = imread(rPath)
      labOut = np.zeros(shape=rnaIn.shape, dtype=np.int32)
      print(rnaIn.shape)
      base, ext = os.path.splitext(filename)
      rLabFile = base + "_lab.tif"
      rLabPath = os.path.join(labPath,rLabFile)
#   print("base=",base," ext=", ext)
      rna = cle.push(rnaIn)
      seg = cle.voronoi_otsu_labeling(rna, spot_sigma = 2, outline_sigma=1)
             
      imsave(rLabPath, seg, check_contrast=False)
```
