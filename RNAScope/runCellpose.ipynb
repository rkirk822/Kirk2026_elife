```python
import os
import numpy as np
from skimage.transform import rescale
from skimage.io import imread, imsave, imshow
from tkinter import Tk
from tkinter import filedialog
from skimage.transform import rescale
from skimage.filters import gaussian

from skimage import io
```

```python
root = Tk()
root.withdraw()
dirName = filedialog.askdirectory()
#print(dirName)
cellPath = dirName + "/cell"
nucPath = dirName + "/nuc"
labPath = dirName + "/cell/cLabs"
ospath = dirName + "/cell/origSize"
if not os.path.exists(cellPath):  
# Create directory for storing the tiff files for cell channels 
   os.mkdir(cellPath)  
files = os.listdir(cellPath)
if not os.path.exists(nucPath):  
# Create directory for storing the tiff files for nuc channels 
   os.mkdir(nucPath)  
if not os.path.exists(labPath):  
# Create directory for storing the tiff files for nuc channels 
   os.mkdir(labPath)  
```

```python
from cellpose import models
from cellpose import io as cpio

#cellPath = "/Users/sachanelson/programming/Jupyter/FISH/dataFromLiwei/cell"
#use_GPU = core.use_gpu()
#print('>>> GPU activated? %d'%use_GPU)

# call logger_setup to have output of cellpose written
from cellpose.io import logger_setup
logger_setup();

model = models.Cellpose(gpu=False, model_type='cyto')
channels = [0,0]
```

```python
for filename in (files):
   if filename.endswith(".tif"):
      fpath = os.path.join(cellPath, filename)
      outpath = os.path.join(labPath, filename)
      print("reading file",fpath)
      img = cpio.imread(fpath)
# might need to play with diameter, pre measure, or use the diameter=None setting
      masks, flows, styles, diams = model.eval(img, diameter=50, do_3D=False, stitch_threshold=0.5)
      print("ran model")
#      fpath2 = cellPath + "/masks"
#      if not os.path.exists(fpath2):  
#         os.mkdir(fpath2)  
      cpio.save_masks(img, masks, flows, outpath, png=False, tif=True, save_flows=True)
   else:
      continue
```
